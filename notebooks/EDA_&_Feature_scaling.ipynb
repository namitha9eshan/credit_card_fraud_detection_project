#### Data Visualization

```python
df.head()
```

```python
#histograms
num_columns = ['NO_OF_CHILD','INCOME', 'AGE', 'FAMILY SIZE','BEGIN_MONTH', 'YEARS_EMPLOYED']

fig, axes = plt.subplots(2,3, figsize = (12, 6))
axes = axes.flatten()
for i, col in enumerate(num_columns):
    sns.histplot(df[col], bins=40, kde=True, ax=axes[i])
    axes[i].set_title(col)

plt.tight_layout()
plt.show()


    
```

```python
#boxplots 
fig , axes = plt.subplots(2, 3, figsize = (12, 6))
axes = axes.flatten()
for i, col in enumerate(num_columns):
    sns.boxplot(x=df[col] , ax=axes[i])
    axes[i].set_title(col)
plt.tight_layout()
plt.show()
```
