## Feature Engineering

```python
print(df.columns.tolist())
```

```python
#drop 'unnamed' , 'ID' & 'FLAG_MOBIL' columns because it does not provide any value
df = df.drop(columns=['Unnamed: 0', 'ID', 'FLAG_MOBIL'])
df.head()
```

```python
df.columns.to_list()
```

```python
df.shape    
```

```python
df['GENDER'].unique()
```

```python
# Handling categorical data --> GENDER

df['GENDER'] = df['GENDER'].map({'M': 0, 'F': 1})
```

```python
df.head()
```

```python
#Handling categorical Data --> CAR
df['CAR'] = df['CAR'].replace({'Y' : 1 , 'N': 0}).astype(int)
```

```python
df.head(2)
```

```python
#Handling Categorical Data --> REALITY  by using one hot encoding 
df['REALITY'] = pd.get_dummies(df['REALITY'], drop_first= True).astype(int)
df.head()
```

```python
df['INCOME_TYPE'].unique()
```

```python
#Since Pensioners and Student has a very low value count I want to merge the data into one --> Other
df['INCOME_TYPE'] = df['INCOME_TYPE'].replace({'Student': 'Other', 'Pensioner': 'Other' })
df['INCOME_TYPE'].unique()
```
